# Setup ddsclient
1. Install ddsclient following instruction from https://github.com/Duke-GCB/DukeDSClient
2. Generate a DDS key following instruction from: <https://github.com/Duke-GCB/DukeDSClient/wiki/Agent-User-Keys-(setup)>
3. Run the following chunk to set dds key permissions and make a backup

```bash
chmod og-w /.ddsclient
cp ~/.ddsclient ~/work/ # to make a backup copy of the key that can be used after container restart
```

# After Container Restart
Run the following chunk after a container restart to restore the ddsclient key.

```bash
# following is necessary after a container restart
cp ~/work/.ddsclient ~/
```
