```r
geneinfo %>%
    pull(seqnames) %>%
    as.character ->
    mychrom

geneinfo %>%
    pull(start) ->
    mystart

geneinfo %>%
    pull(end) ->
    myend

geneinfo %>%
    pull(strand) ->
    mystrand

goi.gr = GRanges(mychrom, IRanges(mystart, myend), strand=mystrand)
```
